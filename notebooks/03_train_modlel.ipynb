## Model Training with cleaned data

this section mainly train the model using cleaned and processed data

review -> tf-idf -> logistic regression -> positive / negative

```python
import sys
from pathlib import Path
from datasets import load_dataset
import pandas as ps

# manipulating the root path
sys.path.append(str(Path.cwd().parent))

from app.ml.preprocess import clean_text
# data loading and processing

dataset = load_dataset('stanfordnlp/imdb')

train_df = dataset['train'].to_pandas()
test_df = dataset['test'].to_pandas()

train_df['clean_text'] = train_df['text'].apply(clean_text)
test_df['clean_text'] = test_df['text'].apply(clean_text)

```

## Experiments

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, recall_score
```

```python
model_lr = Pipeline([
    (
        'tfidf', TfidfVectorizer(
            max_features=50000,
            ngram_range=(1,2)
        )),
        (
            'classifier', LogisticRegression(max_iter=1000)
        )
])

model_lr.fit(
    train_df['clean_text'],
    train_df['label']
)
```

Test the Model

```python
prediction = model_lr.predict(test_df['clean_text'])
```

calculating the accuracy of the model

```python
accuracy = accuracy_score(test_df['label'], prediction)

f1_lr = f1_score(test_df['label'], prediction)

print(accuracy)
```

accuracy report

Linear SVM

```python
model_svm = Pipeline([
    (
        'tfidf', TfidfVectorizer(
            max_features=50000,
            ngram_range=(1,2)
        )),
        (
            'classifier', LinearSVC()
        )
])

model_svm.fit(
    train_df['clean_text'],
    train_df['label']
)
```

```python
prediction_svm = model_svm.predict(test_df['clean_text'])

accuracy_svm = accuracy_score(test_df['label'], prediction_svm)
f1_svm = f1_score(test_df['label'], prediction_svm)

print(accuracy_svm)
```

MNB

```python
model_mnb = Pipeline([
    (
        'tfidf', TfidfVectorizer(
            max_features=50000,
            ngram_range=(1,2)
        )),
        (
            'classifier', MultinomialNB()
        )
])

model_mnb.fit(
    train_df['clean_text'],
    train_df['label']
)
```

```python
prediction_mnb = model_mnb.predict(test_df['clean_text'])

accuracy_mnb = accuracy_score(test_df['label'], prediction_mnb)
f1_mnb = f1_score(test_df['label'], prediction_mnb)

print(accuracy_mnb)
```

## Matrix Compare

```python
import pandas as ps

result = ps.DataFrame({
    "Model": [
        "Logistic Regression",
        "Liner SVM",
        "MNB"
    ],
    "accuracy": [
        accuracy,
        accuracy_svm,
        accuracy_mnb
    ],
    "f1_score":[
        f1_lr,
        f1_svm,
        f1_mnb
    ],
    "recall_score":[
        recall_score(test_df['label'], prediction),
        recall_score(test_df['label'], prediction_svm),
        recall_score(test_df['label'], prediction_mnb)
    ]
})

result
```

## Evaluation Metrics

**Accuracy** represents the proportion of total instances correctly classified by the model across both positive and negative classes

**Precision and recall** Focuses specifically on the accuracy of positive predictions. It measures the proportion of true positive predictions out of all instances predicted as positive: $$\text{Precision} = \frac{\text{True Positives}}{\text{True Positives} + \text{False Positives}}$$

The **F1 Score** represents the harmonic mean of Precision and Recall. It provides a single balanced score when evaluating a model's performance on positive predictions and actual positive cases.Formula: $$\text{F1 Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$

**ROC Curve**: A graph that plots the **True Positive Rate** (Recall) against the **False Positive Rate** across various classification decision thresholds
